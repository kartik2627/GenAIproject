```python
from google.colab import userdata
from pinecone import Pinecone, ServerlessSpec
from sentence_transformers import SentenceTransformer
from transformers import pipeline
```

```python
from google.colab import userdata
PINECONE_API_KEY = userdata.get('api_key')
PINECONE_REGION = "us-east-1"
```

```python
pc = Pinecone(api_key=PINECONE_API_KEY)
```

```python
index_name = "simple-rag-hf"
if index_name not in pc.list_indexes().names():
    pc.create_index(
        name=index_name,
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(cloud="aws", region=PINECONE_REGION)
    )
```

```python
index = pc.Index(index_name)
```

```python
embedder = SentenceTransformer("all-MiniLM-L6-v2")
generator = pipeline("text2text-generation", model="google/flan-t5-base")
```

```python
docs = [
    "Pinecone is a managed vector database for machine learning applications.",
    "RAG combines information retrieval with language generation.",
    "Transformers use attention mechanisms to improve text understanding."
]

doc_embeddings = embedder.encode(docs).tolist()
doc_ids = [f"id{i}" for i in range(len(docs))]
index.upsert(vectors=zip(doc_ids, doc_embeddings))
```

```python
def simple_rag_hf(query):
    query_vec = embedder.encode([query])[0].tolist()
    results = index.query(vector=query_vec, top_k=2, include_metadata=False)
    context = "\n".join([docs[int(match['id'][-1])] for match in results['matches']])
    prompt = f"Context: {context}\n\nQuestion: {query}\nAnswer:"
    response = generator(prompt, max_length=100, do_sample=False)
    return response[0]["generated_text"]
```

```python
print(simple_rag_hf("What does Pinecone do?"))
```
